# Pooling Layer in CNN

## Importing Libraries

```python
import os
os.environ["KERAS_BACKEND"] = "tensorflow"
import keras
```

## Loading Dataset

```python
(X_train,y_train),(X_test,y_test) = keras.datasets.mnist.load_data()
```

```python
X_train.shape,X_test.shape,y_train.shape,y_test.shape
```

## CNN with Pooling Layer

```python
model = keras.Sequential(
    [
        keras.layers.Input(shape=(28,28,1)),

        keras.layers.Conv2D(filters=32, kernel_size=(3,3),padding='valid', activation='relu'),
        keras.layers.MaxPooling2D(pool_size=(2,2), strides=2, padding='valid'),
        
        keras.layers.Conv2D(filters=32, kernel_size=(3,3), activation='relu', padding='valid'),
        keras.layers.MaxPooling2D(pool_size=(2,2), strides=2, padding='valid'),

        keras.layers.Flatten(),

        keras.layers.Dense(units=128, activation='relu'),
        keras.layers.Dense(units=10, activation='softmax'),
    ]
)
```
