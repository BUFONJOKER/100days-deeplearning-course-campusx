# He Normal, Xavier Normal and Uniform Weight Initialization in keras

## Importing Libraries

```python
import os
os.environ['KERAS_BACKEND'] = 'tensorflow'
import keras
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from mlxtend.plotting import plot_decision_regions
from sklearn.datasets import make_moons
```

## Loading the Dataset

```python
X, y = make_moons(100, noise=0.5, random_state=42)
```

```python
plt.scatter(X[:,0],X[:,1],c=y)
plt.show()
```

## Model Building with Glorot Normal Initialization with tanh Activation

```python
model = keras.Sequential(
    [
        keras.layers.Input(shape=(2,)),
        keras.layers.Dense(10, activation='tanh',kernel_initializer=keras.initializers.GlorotNormal()),
        keras.layers.Dense(10, activation='tanh',kernel_initializer=keras.initializers.GlorotNormal()),
        keras.layers.Dense(10, activation='tanh',kernel_initializer=keras.initializers.GlorotNormal()),
        keras.layers.Dense(10, activation='tanh',kernel_initializer=keras.initializers.GlorotNormal()),
        keras.layers.Dense(1, activation='sigmoid'),
    ]
)
model.compile(
    loss=keras.losses.BinaryCrossentropy(),
    optimizer=keras.optimizers.Adam()
)
model.fit(X, y, epochs=100)
plot_decision_regions(X, y, clf=model)
plt.show()
```

## Model Building with Glorot Uniform Initialization with tanh Activation

```python
model = keras.Sequential(
    [
        keras.layers.Input(shape=(2,)),
        keras.layers.Dense(10, activation='tanh',kernel_initializer=keras.initializers.GlorotUniform()),
        keras.layers.Dense(10, activation='tanh',kernel_initializer=keras.initializers.GlorotUniform()),
        keras.layers.Dense(10, activation='tanh',kernel_initializer=keras.initializers.GlorotUniform()),
        keras.layers.Dense(10, activation='tanh',kernel_initializer=keras.initializers.GlorotUniform()),
        keras.layers.Dense(1, activation='sigmoid'),
    ]
)
model.compile(
    loss=keras.losses.BinaryCrossentropy(),
    optimizer=keras.optimizers.Adam()
)
model.fit(X, y, epochs=100)
plot_decision_regions(X, y, clf=model)
plt.show()
```

## Model Building with He Normal Initialization with Relu Activation

```python
model = keras.Sequential(
    [
        keras.layers.Input(shape=(2,)),
        keras.layers.Dense(10, activation='relu',kernel_initializer=keras.initializers.he_normal()),
        keras.layers.Dense(10, activation='relu',kernel_initializer=keras.initializers.he_normal()),
        keras.layers.Dense(10, activation='relu',kernel_initializer=keras.initializers.he_normal()),
        keras.layers.Dense(10, activation='relu',kernel_initializer=keras.initializers.he_normal()),
        keras.layers.Dense(1, activation='sigmoid'),
    ]
)
model.compile(
    loss=keras.losses.BinaryCrossentropy(),
    optimizer=keras.optimizers.Adam()
)
model.fit(X, y, epochs=100)
plot_decision_regions(X, y, clf=model)
plt.show()
```

```python
model = keras.Sequential(
    [
        keras.layers.Input(shape=(2,)),
        keras.layers.Dense(10, activation='relu',kernel_initializer=keras.initializers.he_uniform()),
        keras.layers.Dense(10, activation='relu',kernel_initializer=keras.initializers.he_uniform()),
        keras.layers.Dense(10, activation='relu',kernel_initializer=keras.initializers.he_uniform()),
        keras.layers.Dense(10, activation='relu',kernel_initializer=keras.initializers.he_uniform()),
        keras.layers.Dense(1, activation='sigmoid'),
    ]
)
model.compile(
    loss=keras.losses.BinaryCrossentropy(),
    optimizer=keras.optimizers.Adam()
)
model.fit(X, y, epochs=100)
plot_decision_regions(X, y, clf=model)
plt.show()
```
