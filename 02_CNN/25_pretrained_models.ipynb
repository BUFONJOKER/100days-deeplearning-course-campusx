# Pretrained models in CNN - ImageNET Dataset - ILSVRC

```python
import os
os.environ["KERAS_BACKEND"] = "tensorflow"
import keras
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

```python
alexnet_model = keras.Sequential(
    [
        keras.layers.Input(shape=(227,227,3)),
        # 1
        keras.layers.Conv2D(filters=96, kernel_size=(11,11), strides=4, activation='relu', kernel_initializer= 'he_normal'),
        keras.layers.MaxPool2D(pool_size=(3,3), strides=2),

        # 2
        keras.layers.Conv2D(filters=256, kernel_size=(5,5),strides=1, activation='relu', kernel_initializer= 'he_normal', padding='same'),
        keras.layers.MaxPool2D(pool_size=(3,3), strides=2),

        # 3
        keras.layers.Conv2D(filters=384, kernel_size=(3,3), strides=1, activation='relu', padding='same', kernel_initializer= 'he_normal'),

        # 4
        keras.layers.Conv2D(filters=384, kernel_size=(3,3), strides=1, activation='relu', padding='same', kernel_initializer= 'he_normal'),

        # 5
        keras.layers.Conv2D(filters=256, kernel_size=(3,3), strides=1, activation='relu', padding='same', kernel_initializer= 'he_normal'),
        keras.layers.MaxPool2D(pool_size=(3,3), strides=2),


        keras.layers.Flatten(),

        keras.layers.Dense(units=4096, activation='relu'),
        keras.layers.Dense(units=4096, activation='relu'),
        keras.layers.Dense(units=1000, activation='softmax')

    ]
)
```

```python
alexnet_model.summary()
```

## Classify ImageNet classes with ResNet50

```python
from keras.applications.resnet50 import ResNet50
model = ResNet50(weights='imagenet')
```

```python
image_bat = 'bat.jpg'
image_lion = 'lion.jpg'
image_printer = 'printer.jpg'
```

```python
img = keras.utils.load_img(image_printer, target_size=(224,224))
```

```python
from keras.applications.resnet50 import preprocess_input
x = keras.utils.img_to_array(img)
x = np.expand_dims(x, axis=0)
x = preprocess_input(x)
```

```python
preds = model.predict(x)
```

```python
from keras.applications.resnet50 import decode_predictions
print('Predicted')
print(decode_predictions(preds, top=3)[0])
```
