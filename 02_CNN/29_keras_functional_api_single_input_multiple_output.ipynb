# Keras Functional API Example for Non-Linear Neural Networks ( Single Input with Multiple Outputs )

```python
import os
os.environ["KERAS_BACKEND"] = "tensorflow"
import keras
```

```python
x = keras.layers.Input(shape=(32,))
hidden_1 = keras.layers.Dense(units=128, activation='relu')(x)
hidden_2 = keras.layers.Dense(units=64, activation='relu')(hidden_1)
```

```python
output_1 = keras.layers.Dense(units=1, activation='linear')(hidden_2)
output_2 = keras.layers.Dense(units=1, activation='sigmoid')(hidden_2)
```

```python
model = keras.models.Model(inputs=x, outputs=[output_1, output_2])
```

```python
model.summary()
```

```python
keras.utils.plot_model(model, show_shapes=True)
```
